```python
# Create date-based features

df_clean["date_year"] = df_clean["date"].dt.year

df_clean["quarter"] = df_clean["date"].dt.quarter

df_clean["week_of_year"] = df_clean["date"].dt.isocalendar().week.astype("Int64")

df_clean["day"] = df_clean["date"].dt.day

df_clean["is_weekend"] = (
    df_clean["date"].dt.dayofweek >= 5
).astype("Int64")

print("Date features created successfully.")

display(
    df_clean[
        [
            "date",
            "date_year",
            "quarter",
            "week_of_year",
            "day",
            "is_weekend"
        ]
    ].head()
)
```

```python
def classify_aqi_severity(category):
    if category in ["Good", "Satisfactory"]:
        return "Better"
    elif category == "Moderate":
        return "Moderate"
    elif category in ["Poor", "Very Poor", "Severe"]:
        return "Poor"
    else:
        return "Unknown"


df_clean["aqi_severity"] = df_clean["aqi_category"].apply(
    classify_aqi_severity
)

print("AQI severity created.")

display(
    df_clean["aqi_severity"]
    .value_counts()
)
```

```python
pollutant_text = (
    df_clean["prominent_pollutant"]
    .fillna("")
    .str.upper()
)

df_clean["has_pm25"] = pollutant_text.str.contains(
    "PM2.5",
    regex=False
).astype("Int64")

df_clean["has_pm10"] = pollutant_text.str.contains(
    "PM10",
    regex=False
).astype("Int64")

df_clean["has_o3"] = pollutant_text.str.contains(
    "O3",
    regex=False
).astype("Int64")

df_clean["has_no2"] = pollutant_text.str.contains(
    "NO2",
    regex=False
).astype("Int64")

df_clean["has_so2"] = pollutant_text.str.contains(
    "SO2",
    regex=False
).astype("Int64")

df_clean["has_co"] = pollutant_text.str.contains(
    "CO",
    regex=False
).astype("Int64")

print("Pollutant flags created.")
```

```python
city_gap_results = []

for (state, city), group in df_clean.groupby(["state", "city"]):

    dates = pd.Series(
        group["date"].drop_duplicates().sort_values()
    )

    if len(dates) < 2:
        city_gap_results.append({
            "state": state,
            "city": city,
            "max_consecutive_missing_days": 0,
            "missing_days": 0
        })
        continue

    full_range = pd.date_range(
        start=dates.min(),
        end=dates.max(),
        freq="D"
    )

    missing_dates = full_range.difference(dates)

    if len(missing_dates) == 0:
        max_gap = 0
    else:
        missing_series = pd.Series(missing_dates)

        gap_groups = (
            missing_series
            .diff()
            .dt.days
            .ne(1)
            .cumsum()
        )

        max_gap = (
            missing_series
            .groupby(gap_groups)
            .size()
            .max()
        )

    city_gap_results.append({
        "state": state,
        "city": city,
        "max_consecutive_missing_days": int(max_gap),
        "missing_days": int(len(missing_dates))
    })

city_gaps = pd.DataFrame(city_gap_results)

display(city_gaps)
```

```python
# ------------------------------------------------------------
# A. CITY COVERAGE TABLE
# ------------------------------------------------------------

city_coverage = (
    df_clean
    .groupby(["state", "city"], as_index=False)
    .agg(
        records=("aqi", "count"),
        first_date=("date", "min"),
        last_date=("date", "max"),
        years_covered=("year", "nunique"),
        avg_stations_reporting=("stations_reporting", "mean"),
        latitude=("latitude", "first"),
        longitude=("longitude", "first")
    )
)

# Calendar coverage
city_coverage["calendar_days"] = (
    city_coverage["last_date"] - city_coverage["first_date"]
).dt.days + 1

city_coverage["missing_days"] = (
    city_coverage["calendar_days"] - city_coverage["records"]
)

city_coverage["coverage_pct"] = (
    city_coverage["records"]
    / city_coverage["calendar_days"]
    * 100
)


# ------------------------------------------------------------
# Longest internal missing-day gap
# ------------------------------------------------------------

def longest_missing_gap(group):
    dates = (
        group["date"]
        .drop_duplicates()
        .sort_values()
    )

    if len(dates) <= 1:
        return 0

    gaps = dates.diff().dt.days - 1

    return int(gaps.max())


longest_gaps = (
    df_clean
    .groupby(["state", "city"])
    .apply(longest_missing_gap, include_groups=False)
    .reset_index(name="longest_gap")
)

city_coverage = city_coverage.merge(
    longest_gaps,
    on=["state", "city"],
    how="left"
)


# ------------------------------------------------------------
# Coverage level
# ------------------------------------------------------------

city_coverage["coverage_level"] = pd.cut(
    city_coverage["records"],
    bins=[-1, 90, 180, 364, float("inf")],
    labels=["Limited", "Low", "Medium", "High"]
)


# ------------------------------------------------------------
# Location field
# ------------------------------------------------------------

city_coverage["location"] = (
    city_coverage["city"].astype(str)
    + ", "
    + city_coverage["state"].astype(str)
)


# ------------------------------------------------------------
# Round numeric fields
# ------------------------------------------------------------

city_coverage["avg_stations_reporting"] = (
    city_coverage["avg_stations_reporting"].round(2)
)

city_coverage["coverage_pct"] = (
    city_coverage["coverage_pct"].round(2)
)

city_coverage["latitude"] = city_coverage["latitude"].round(6)
city_coverage["longitude"] = city_coverage["longitude"].round(6)


# ------------------------------------------------------------
# Reorder columns
# ------------------------------------------------------------

city_coverage = city_coverage[
    [
        "state",
        "city",
        "location",
        "latitude",
        "longitude",
        "records",
        "first_date",
        "last_date",
        "years_covered",
        "calendar_days",
        "missing_days",
        "coverage_pct",
        "longest_gap",
        "coverage_level",
        "avg_stations_reporting"
    ]
]


# ============================================================
# B. SOURCE METADATA TABLE
# ============================================================

source_metadata = pd.DataFrame({
    "source": [
        "UEI",
        "JanVayu"
    ],

    "period_start": [
        pd.Timestamp("2016-01-01"),
        pd.Timestamp("2026-01-01")
    ],

    "period_end": [
        pd.Timestamp("2025-12-31"),
        pd.Timestamp("2026-09-17")
    ],

    "period_label": [
        "2016-2025",
        "2026 YTD"
    ],

    "source_type": [
        "Historical dataset",
        "Historical dataset"
    ],

    "stations_total_available": [
        "No",
        "Yes"
    ]
})

print("CITY COVERAGE")
print("=" * 60)

print("Rows:", city_coverage.shape[0])
print("Columns:", city_coverage.shape[1])

display(city_coverage.head(10))

print("\nCoverage levels:")
display(
    city_coverage["coverage_level"]
    .value_counts()
    .sort_index()
)

print("\nMissing values:")
display(
    city_coverage.isna()
    .sum()
    .sort_values(ascending=False)
    .to_frame("missing_values")
)


print("\n\nSOURCE METADATA")
print("=" * 60)

display(source_metadata)

print("\nMissing values:")
display(
    source_metadata.isna()
    .sum()
    .to_frame("missing_values")
)
```
